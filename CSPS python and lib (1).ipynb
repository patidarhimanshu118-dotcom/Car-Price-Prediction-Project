# 1. Handling Missing Values in selling_price
Identify the maximum and minimum values of the selling_price column.    
Compute the average of these two values.                                         
Use this calculated average to fill all missing values in the selling_price column.   

```python
max_price = data["selling_price"].max()
max_price
```

```python
min_price = data["selling_price"].min()
min_price
```

```python
avg = (max_price + min_price) / 2
avg
```

```python
data["selling_price"]=data["selling_price"].fillna(avg)
```

```python
data["selling_price"].isnull().any()
```
